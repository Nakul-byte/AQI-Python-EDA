2020 analysis

```python
df_2020=df_final[df_final['year_month'].dt.year==2020]
df_2020.head(1)
```

```python
def get_period (month):
    if month<=3:
        return 'pre-lockdown'
    elif month<=6:
        return 'lockdown'
    else:
        return 'post-lockdown'

df_2020['period']=df_2020['year_month'].dt.month.apply(get_period)
```

```python
df_2020['period'].value_counts()
```

```python
df_2020_compare=df_2020[df_2020['period']!='post-lockdown']
df_2020_compare.shape
```

GRAPHICAL REPRESENTATION OF PRE-LOCKDOWN AND POST-LOCKDOWN

```python
df_2020_compare.groupby(['period','city'])['aqi'].mean().unstack('city').plot(kind='bar', figsize=(10,6))
plt.title('Average AQI by City: Pre-Lockdown vs Lockdown (Jan-Jun 2020)')
plt.ylabel('Average AQI')
plt.xlabel('Period')
plt.xticks(rotation=0)
plt.show()
```

```python
period_avg = df_2020_compare.groupby(['period','city'])['aqi'].mean().unstack('city')
period_avg
```

```python
lockdown_values = period_avg.loc['lockdown']
prelockdown_values = period_avg.loc['pre-lockdown']
```

```python
drop=lockdown_values-prelockdown_values
per_change=(drop/prelockdown_values)*100
per_change.round(1)
```

Chennai showed a slight AQI increase during lockdown (+3.9%), unlike the other 4 cities — this may relate to coastal weather patterns or monitoring gaps in the dataset, and is noted as an area for further investigation

* Cyclone Amphan hit the Bay of Bengal in mid-May 2020, which could have affected regional air/weather patterns right in the middle of your lockdown window

POLLUTION COMPONENTS BEFORE AND AFTER LOCKDOWN

```python
pollutant_cols = ['pm2.5','pm10','no','no2','nox','co','so2','o3','benzene','toluene']
pollutant_avg = df_2020_compare.groupby(['period','city'])[pollutant_cols].mean().unstack('city')
pollutant_avg
```

```python
lockdown_vals = pollutant_avg.loc['lockdown']
prelockdown_vals = pollutant_avg.loc['pre-lockdown']

drop = lockdown_vals - prelockdown_vals
percent_drop = (drop / prelockdown_vals) * 100
percent_drop.round(1).sort_values().unstack('city')
```

O3 (ozone) rising in both Delhi and Chennai during lockdown is a genuinely known, documented phenomenon — ozone often increases when NOx drops, because NOx normally reacts with and consumes ground-level ozone. Fewer vehicles → less NOx → less ozone consumption → ozone rises. This is real atmospheric chemistry
